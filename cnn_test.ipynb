```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model

image_path = './AugmentedAlzheimerDataset/MildDemented/886ae3a3-c7fe-40db-a8e1-579dbe2abf6d.jpg'

image_processed = cv2.imread(image_path)
resized_image = cv2.resize(image_processed, (128,128))

plt.imshow(resized_image)
plt.show()


adjusted_image = np.expand_dims(resized_image, axis=-1)
adjusted_image = adjusted_image/255

image_to_array = np.array(adjusted_image)

model = load_model('cnn_alzheimer_diagnosis.h5')

model.predict(image_to_array)
```
